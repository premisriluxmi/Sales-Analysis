# Sales Analysis

## Objective

To analyze a collection of sales values using Python and calculate the total sales, average sales, highest sale, and lowest sale.

```python
# Sales data

sales = [
    12500, 18000, 9500, 22000, 15750,
    19800, 27500, 14300, 16500, 31000,
    11200, 24500, 17800, 29000, 13500,
    20500, 32000, 15200, 26000, 19000
]

sales
```

```python
total_sales = sum(sales)
total_sales
```

```python
total_sales = average(sales)
total_sales
```
